```python
import pandas as pd

FILE_PATH = r"C:\Users\admin\Desktop\student-work\Student project\Students data.csv"

data = pd.read_csv(FILE_PATH)

male_scores = data[data["gender"] == "male"]["GPA"]
female_scores = data[data["gender"] == "female"]["GPA"]

male_count = 0
male_total = 0
for score in male_scores:
    male_count += 1
    male_total += score

female_count = 0
female_total = 0
for score in female_scores:
    female_count += 1
    female_total += score

print("Male students:", male_count)
print("Female students:", female_count)
print("Average male grade point average:", male_total / male_count)
print("Average female grade point average:", female_total / female_count)
```
