```python
# Parent class (base class)5
#Inheritance
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."

    def sleep(self):
        return f"{self.name} is sleeping."

    def speak(self):
        return f"{self.name} makes a sound."


# Child class (subclass) inheriting from Animal
class Cat(Animal):
    def speak(self):    # Overriding the parent's method
        return f"{self.name} says Meow!"

    def purr(self):     # New method only in Cat
        return f"{self.name} is purring."


class Dog(Animal):
    def __init__(self, name, breed):    # Extending __init__
        super().__init__(name)          # Call parent's __init__
        self.breed = breed

    def speak(self):
        return f"{self.name} says Woof!"

    def fetch(self):
        return f"{self.name} the {self.breed} is fetching the ball."


cat = Cat("Whiskers")
dog = Dog("Rex", "Labrador")

print(cat.eat())      # Whiskers is eating. (inherited)
print(cat.speak())    # Whiskers says Meow! (overridden)
print(cat.purr())     # Whiskers is purring. (new method)

print(dog.speak())    # Rex says Woof!
print(dog.fetch())    # Rex the Labrador is fetching the ball.
print(dog.breed)      # Labrador
```

```python
#incapsulation
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner            # public
        self._account_type = "Savings"  # protected (convention)
        self.__balance = balance      # private (name-mangled)

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter method with validation
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew ${amount}. New balance: ${self.__balance}")
        else:
            print("Invalid withdrawal amount.")


account = BankAccount("Alice", 1000)
print(account.owner)              # Alice
print(account._account_type)      # Savings (accessible but discouraged)

# print(account.__balance)        # AttributeError! It's private
print(account.get_balance())      # 1000 (proper way)

account.deposit(500)              # Deposited $500. New balance: $1500
account.withdraw(200)             # Withdrew $200. New balance: $1300

# Name mangling - Python renames __balance to _BankAccount__balance
print(account._BankAccount__balance)  # 1300 (possible but bad practice)
```

```python
#Abstraction
from abc import ABC, abstractmethod

# Abstract Base Class - cannot be instantiated directly
class Payment(ABC):
    def __init__(self, amount):
        self.amount = amount

    @abstractmethod
    def pay(self):    # Every subclass MUST implement this
        pass

    def receipt(self):   # Concrete method - can be used as-is
        return f"Payment of ${self.amount} processed."


class CreditCard(Payment):
    def pay(self):
        return f"Paid ${self.amount} via Credit Card."

class PayPal(Payment):
    def pay(self):
        return f"Paid ${self.amount} via PayPal."


# payment = Payment(100)   # ERROR - can't instantiate abstract class

cc = CreditCard(100)
pp = PayPal(50)

print(cc.pay())        # Paid $100 via Credit Card.
print(pp.pay())        # Paid $50 via PayPal.
print(cc.receipt())    # Payment of $100 processed.
```

```python
#polymorphism
class Shape:
    def area(self):
        return "Area not defined"

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def area(self):
        return 0.5 * self.base * self.height


# Same function call works for all shapes!
shapes = [Circle(5), Square(4), Triangle(6, 3)]

for shape in shapes:
    print(f"{type(shape).__name__} area: {shape.area()}")

# Output:
# Circle area: 78.5
# Square area: 16
# Triangle area: 9.0
```

```python
#overriding
class Calculator:
    def add(self, a, b=0, c=0):   # Optional parameters
        return a + b + c

calc = Calculator()
print(calc.add(2))          # 2
print(calc.add(2, 3))       # 5
print(calc.add(2, 3, 4))    # 9
```

```python
#class methods
class Employee:
    company = "TechCorp"      # class attribute
    count = 0                 # counts all employees

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.count += 1

    # Instance method - works with instance (self)
    def describe(self):
        return f"{self.name} earns ${self.salary}"

    # Class method - works with class (cls)
    @classmethod
    def get_company(cls):
        return cls.company

    @classmethod
    def from_string(cls, data_string):
        """Alternative constructor"""
        name, salary = data_string.split("-")
        return cls(name, int(salary))

    # Static method - doesn't need self or cls, just a utility
    @staticmethod
    def is_workday(day):
        return day.lower() not in ["saturday", "sunday"]


e1 = Employee("Alice", 50000)
e2 = Employee.from_string("Bob-60000")   # uses class method as constructor

print(e1.describe())                     # Alice earns $50000
print(e2.describe())                     # Bob earns $60000
print(Employee.get_company())            # TechCorp
print(Employee.is_workday("Monday"))     # True
print(Employee.is_workday("Sunday"))     # False
print(Employee.count)                    # 2
```

```python
#special magic
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    # Called by print() and str()
    def __str__(self):
        return f"Book: {self.title}"

    # Called by repr() and in interactive shell
    def __repr__(self):
        return f"Book('{self.title}', {self.pages})"

    # Called by len()
    def __len__(self):
        return self.pages

    # Called by ==
    def __eq__(self, other):
        return self.title == other.title and self.pages == other.pages

    # Called by <
    def __lt__(self, other):
        return self.pages < other.pages


b1 = Book("Python 101", 300)
b2 = Book("OOP Guide", 200)

print(b1)              # Book: Python 101
print(repr(b1))        # Book('Python 101', 300)
print(len(b1))         # 300
print(b1 == b2)        # False
print(b2 < b1)         # True (200 < 300)
```

```python
#inheritance types
# 1. Single Inheritance
class A:
    pass
class B(A):
    pass

# 2. Multiple Inheritance
class Flyer:
    def fly(self):
        return "I can fly!"

class Swimmer:
    def swim(self):
        return "I can swim!"

class Duck(Flyer, Swimmer):   # Inherits from TWO parents
    pass

d = Duck()
print(d.fly())    # I can fly!
print(d.swim())   # I can swim!

# 3. Multilevel Inheritance
class Grandparent:
    def greet(self):
        return "Hello from Grandparent"

class Parent(Grandparent):
    pass

class Child(Parent):    # Inherits through Parent
    pass

c = Child()
print(c.greet())   # Hello from Grandparent
```

```python
#super fucntion
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def describe(self):
        return f"Brand: {self.brand}"

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)   # Call Vehicle's __init__
        self.model = model

    def describe(self):
        # Call parent's describe, then extend it
        parent_desc = super().describe()
        return f"{parent_desc}, Model: {self.model}"

car = Car("Toyota", "Corolla")
print(car.describe())   # Brand: Toyota, Model: Corolla
```

```python
#getters and setters
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):              # getter
        return self._celsius

    @celsius.setter
    def celsius(self, value):       # setter
        if value < -273.15:
            raise ValueError("Temperature below absolute zero!")
        self._celsius = value

    @property
    def fahrenheit(self):           # computed property
        return (self._celsius * 9/5) + 32


t = Temperature(25)
print(t.celsius)         # 25
print(t.fahrenheit)      # 77.0

t.celsius = 30           # uses setter
print(t.fahrenheit)      # 86.0

# t.celsius = -300       # raises ValueError
```

```python
#duck typing polymorphism
class Duck:
    def sound(self):
        return "Quack!"

class Person:
    def sound(self):
        return "Hello!"

class Car:
    def sound(self):
        return "Vroom!"

# This function doesn't care about the type, only that .sound() exists
def make_sound(obj):
    print(obj.sound())

make_sound(Duck())    # Quack!
make_sound(Person())  # Hello!
make_sound(Car())     # Vroom!
```

```python
#wrapping them up
from abc import ABC, abstractmethod


# ABSTRACT BASE CLASS
class Employee(ABC):
    company = "TechCorp"

    def __init__(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self._salary = salary       # protected

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value < 0:
            raise ValueError("Salary cannot be negative")
        self._salary = value

    @abstractmethod
    def role(self):
        pass

    def __str__(self):
        return f"{self.name} (ID: {self.emp_id})"


# SUBCLASS 1
class Developer(Employee):
    def __init__(self, name, emp_id, salary, language):
        super().__init__(name, emp_id, salary)
        self.language = language

    def role(self):
        return f"{self.name} is a {self.language} Developer."

    def code(self):
        return f"{self.name} is writing {self.language} code."


# SUBCLASS 2
class Manager(Employee):
    def __init__(self, name, emp_id, salary, team_size):
        super().__init__(name, emp_id, salary)
        self.team_size = team_size

    def role(self):
        return f"{self.name} manages a team of {self.team_size}."


# POLYMORPHISM IN ACTION
employees = [
    Developer("Alice", 101, 80000, "Python"),
    Manager("Bob", 102, 95000, 5),
    Developer("Charlie", 103, 75000, "JavaScript"),
]

for emp in employees:
    print(emp.role())
    print(f"  Salary: ${emp.salary}")
    print()

# ENCAPSULATION DEMO
alice = employees[0]
alice.salary = 85000
print(f"Alice's new salary: ${alice.salary}")

# try:
#     alice.salary = -100    # ValueError
# except ValueError as e:
#     print(e)
```
