```python
FILE_PATH = r"C:\Users\admin\Desktop\student-work\Student project\Students data.csv"

male_scores = []
female_scores = []

file = open(FILE_PATH)
file.readline()                      # skip the column names

for line in file:
    parts = line.strip().split(",")
    gender = parts[2]                # third column
    score = float(parts[4])          # fifth column (grade point average)

    if gender == "male":
        male_scores.append(score)
    else:
        female_scores.append(score)

file.close()

print("Male students:", len(male_scores))
print("Female students:", len(female_scores))
print("Average male grade point average:", sum(male_scores) / len(male_scores))
print("Average female grade point average:", sum(female_scores) / len(female_scores))
```
