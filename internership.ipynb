# Data cleaning project

```python
def clean_ages(data):
    cleaned = []
    for item in data:
        try:
            cleaned.append(int(item))
        except ValueError:
            continue  # skip bad values like "N/A" or ""
    return cleaned

ages_raw = ["23", "45", "N/A", "31", "", "29", "abc"]
print("Cleaned ages:", clean_ages(ages_raw))
```

# Word frequency Project with manual counting

```python
def word_frequency(text):
    # replace punctuation with spaces, lowercase everything
    cleaned_text = ""
    for c in text:
        if c.isalnum() or c.isspace():
            cleaned_text += c.lower()
        else:
            continue

    words = cleaned_text.split()

    counts = {}
    for w in words:
        counts[w] = counts.get(w, 0) + 1
    return counts

sample_text = "The cat sat on the mat. The cat was happy!"
print("Word frequency:", word_frequency(sample_text))
```

# Word frequency project with counter

```python
from collections import Counter

def word_frequency_counter(text):
    cleaned_text = ""
    for c in text:
        if c.isalnum() or c.isspace():
            cleaned_text +=c.lower()
        else:
            continue
    words = cleaned_text.split()
    return dict(Counter(words))

print("Word frequency (counter):", word_frequency_counter(sample_text))
```

# Simple statistics

```python
def mean(nums):
    return sum(nums) / len(nums)

def median(nums):
    s = sorted(nums)
    n = len(s)
    mid = n // 2
    if n % 2 == 0:
        return (s[mid - 1] + s[mid]) / 2
    return s[mid]

def stddev(nums):
    avg = mean(nums)
    diffs = [(x - avg) ** 2 for x in nums]
    return (sum(diffs) / len(nums)) ** 0.5

sample_nums = [4, 8, 6, 5, 3, 2, 8, 9, 2, 5]
print("Mean:", mean(sample_nums))
print("Median:", median(sample_nums))
print("Stddev:", stddev(sample_nums))
```

# Operations on CSV files

```python
import csv

sample_csv_path = "people.csv"
with open(sample_csv_path, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "age", "city"])
    writer.writerow(["Alice", "30", "Nairobi"])
    writer.writerow(["Bob", "25", "Kigali"])
    writer.writerow(["Carol", "35", "Nairobi"])
    writer.writerow(["Dan", "28", "Kampala"])

def read_people_csv(path):
    with open(path, "r", newline="") as f:
        reader = csv.DictReader(f)
        return list(reader)

people = read_people_csv(sample_csv_path)
average_age = mean([int(p["age"]) for p in people])
print("Average age:", average_age)

# how many people per city
city_counts = {}
for p in people:
    city = p["city"]
    city_counts[city] = city_counts.get(city, 0) + 1
print("People per city:", city_counts)
```

# Classes implementation

```python
class Dataset:
    def __init__(self, values):
        self.values = values

    def mean(self):
        return sum(self.values) / len(self.values)

    def max(self):
        return max(self.values)

    def min(self):
        return min(self.values)

    def normalize(self):
        lo, hi = self.min(), self.max()
        if hi == lo:
            return [0.0 for _ in self.values]
        return [(v - lo) / (hi - lo) for v in self.values]

    def __len__(self):
        return len(self.values)

    def __repr__(self):
        return f"Dataset({self.values})"

ds = Dataset([10, 20, 30, 40, 50])
print("Dataset:", ds)
print("Length:", len(ds))
print("Mean:", ds.mean())
print("Normalized:", ds.normalize())
```

# Inheritance

```python
import json

class DataLoader:
    def load(self):
        raise NotImplementedError("Subclasses must implement load()")

class CSVLoader(DataLoader):
    def __init__(self, path):
        self.path = path

    def load(self):
        with open(self.path, "r", newline="") as f:
            reader = csv.DictReader(f)
            return list(reader)

class JSONLoader(DataLoader):
    def __init__(self, path):
        self.path = path

    def load(self):
        with open(self.path, "r") as f:
            return json.load(f)

sample_json_path = "people.json"
with open(sample_json_path, "w") as f:
    json.dump([{"name": "Eve", "age": 40, "city": "Kigali"}], f)

loaders = [CSVLoader(sample_csv_path), JSONLoader(sample_json_path)]
print("Polymorphic loading:")
for loader in loaders:
    # doesn't matter which loader this is, .load() works the same
    print(f"  {type(loader).__name__} loaded:", loader.load())
```

# Setters, getters, and properties on classes

```python
class SensorReading:
    def __init__(self, value):
        self.value = value

    @property
    def value(self):
        return self._value

    @value.setter
    def value(self, new_value):
        if not (0 <= new_value <= 100):
            raise ValueError("Sensor reading must be between 0 and 100")
        self._value = new_value

sensor = SensorReading(45)
print("Initial sensor value:", sensor.value)

sensor.value = 99
print("Updated sensor value:", sensor.value)

try:
    sensor.value = 150  # should fail
except ValueError as e:
    print("Caught expected error:", e)
```

# A program about finding the average score of a student

```python
class Student:
    def __init__(self, name, scores=None):
        self.name = name
        self.scores = scores if scores is not None else []

    def average(self):
        if not self.scores:
            return 0
        return sum(self.scores) / len(self.scores)

    def __repr__(self):
        return f"Student({self.name}, avg={self.average():.1f})"

class Classroom:
    def __init__(self):
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def top_student(self):
        return max(self.students, key=lambda s: s.average())

    def average_score(self):
        if not self.students:
            return 0
        return sum(s.average() for s in self.students) / len(self.students)

classroom = Classroom()
classroom.add_student(Student("Amara", [80, 90, 85]))
classroom.add_student(Student("Liam", [70, 75, 60]))
classroom.add_student(Student("Zara", [95, 92, 98]))

print("Top student:", classroom.top_student())
print("Classroom average:", classroom.average_score())
```

# Mini ML pipeline program

```python

class Preprocessor:
    def transform(self, data):
        max_val = max(data) if data else 1
        return [x / max_val for x in data]

class Model:
    def __init__(self):
        self.is_fitted = False

    def fit(self, data):
        self.is_fitted = True
        print("  Model fitted on data:", data)

    def predict(self, data):
        if not self.is_fitted:
            raise RuntimeError("Model must be fitted before predicting")
        return [x * 2 for x in data]

class Pipeline:
    def __init__(self, preprocessor, model):
        self.preprocessor = preprocessor
        self.model = model

    def run(self, data):
        transformed = self.preprocessor.transform(data)
        self.model.fit(transformed)
        return self.model.predict(transformed)

pipeline = Pipeline(Preprocessor(), Model())
raw_data = [10, 20, 30, 40]
result = pipeline.run(raw_data)
print("Pipeline result:", result)
```

## Data Cleaning

### fixing, removing, incomplete, incorrect, or irrelevant data 

```python
import pandas as pd

df = pd.read_csv("data.csv")

# drop unnecessary column (keep Legendary so we can convert it below)
df = df.drop(columns=["No"])
print(df.to_string())

```

```python
# standardize text
df["Name"] = df["Name"].str.lower()

# fix data types
df["Legendary"] = df["Legendary"].astype(bool)
print(df.to_string())
```

```python
# fill missing values (single-type Pokémon get "None")
df = df.fillna({"Type2": "None"})
print(df.to_string())
```

```python
# fix inconsistent values: make type casing uniform
df["Type1"] = df["Type1"].str.upper()
df["Type2"] = df["Type2"].str.upper()
print(df.to_string())
```

```python
# remove duplicates
df = df.drop_duplicates()
print(df.to_string())
```
