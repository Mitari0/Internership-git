```python
class Bank:
    def __init__(self, bankname):
        self.bankname = bankname

    def Interest(self, principal, rate, time):
        self.principal = principal
        self.rate = rate
        self.time = time
        self.I = self.principal * (self.rate * 0.01) * (self.time / 12)
        return self.I

    def Amount(self):
        return self.principal + self.I


class Branch(Bank):
    def __init__(self, bankname, branchname):
        super().__init__(bankname)   # calls Bank's __init__ to set up bankname
        self.branchname = branchname

branch = Branch("Urwego", "Kigali Branch")

print(branch.bankname)                          # Urwego (inherited)
print(branch.branchname)                        # Kigali Branch (its own)
print(branch.Interest(1000, 20, 24))             # inherited method works
print(branch.Amount())                           # inherited method works
```

```python
class Company:
    def __init__(self, company_name):
        self._company_name = company_name
        self._revenue = 0
        self._expenses = 0

    # getter for revenue
    def get_revenue(self):
        return self._revenue

    # setter for revenue
    def set_revenue(self, revenue):
        if revenue < 0:
            print("Revenue can't be negative")
        else:
            self._revenue = revenue

    # getter for expenses
    def get_expenses(self):
        return self._expenses

    # setter for expenses
    def set_expenses(self, expenses):
        if expenses < 0:
            print("Expenses can't be negative")
        else:
            self._expenses = expenses

    # calculate profit or loss
    def calculate_result(self):
        result = self._revenue - self._expenses
        if result > 0:
            return f"Profit of {result}"
        elif result < 0:
            return f"Loss of {abs(result)}"
        else:
            return "No profit, no loss (break even)"


# get input from the user
name = input("Enter company name: ")
company = Company(name)

revenue = float(input("Enter total revenue: "))
expenses = float(input("Enter total expenses: "))

# using setters
company.set_revenue(revenue)
company.set_expenses(expenses)

# using getters
print(f"\n{company._company_name}'s revenue: {company.get_revenue()}")
print(f"{company._company_name}'s expenses: {company.get_expenses()}")
print(company.calculate_result())
```

```python
import pandas as pd

path = r"C:\Users\admin\OneDrive - United States International University (USIU)\Desktop\DSA\traffic-accident-severity-prediction-main\traffic-accident-severity-prediction-1\notebooks\cleaned.csv"

df = pd.read_csv(path)

before_count = len(df)   # count rows BEFORE cleanup
print(f"Rows before cleanup: {before_count}")

# remove duplicate rows
df = df.drop_duplicates()

# remove rows that are completely empty
df = df.dropna(how="all")

# strip extra whitespace from text columns
df = df.apply(lambda col: col.str.strip() if col.dtype == "object" else col)

after_count = len(df)   # count rows AFTER cleanup
print(f"Rows after cleanup: {after_count}")
print(f"Rows removed: {before_count - after_count}")

# save the cleaned version back
df.to_csv(path, index=False)
```
