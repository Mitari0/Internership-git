```python
# importing required libraries
import pandas as pd
#importing student data from csv file

student_data = pd.read_csv(r"C:\Users\admin\OneDrive - United States International University (USIU)\Desktop\DSA\students (1).csv")
print(student_data)

#finding the average age of students based on gender
def Average_age(student_data):
    average_age = student_data.groupby('Gender').mean('Age')
    return average_age
print(Average_age(student_data))

#counting the number of students based on gender
def counting_gender(student_data):
    gender_count = student_data['Gender'].value_counts()
    return gender_count
print(counting_gender(student_data))
```
