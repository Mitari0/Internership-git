## Word frequency Project with manual counting

```python
def word_frequency(text):
    # replace punctuation with spaces, lowercase everything
    cleaned_text = ""
    for c in text:
        if c.isalnum() or c.isspace():
            cleaned_text += c.lower()
        else:
            continue

    words = cleaned_text.split()

    counts = {}
    for w in words:
        counts[w] = counts.get(w, 0) + 1
    return counts

sample_text = "The cat sat on the mat. The cat was happy!"
print("Word frequency:", word_frequency(sample_text))
```

## word frequency with counter

```python
from collections import Counter

def word_frequency_counter(text):
    cleaned_text = ""
    for c in text:
        if c.isalnum() or c.isspace():
            cleaned_text +=c.lower()
        else:
            continue
    words = cleaned_text.split()
    return dict(Counter(words))

print("Word frequency (counter):", word_frequency_counter(sample_text))
```
