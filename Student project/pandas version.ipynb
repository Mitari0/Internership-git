```python
import pandas as pd

FILE_PATH = r"C:\Users\admin\Desktop\student-work\Student project\Students data.csv"

data = pd.read_csv(FILE_PATH)

male_count = 0
male_total = 0
female_count = 0
female_total = 0

# Go through the table one row at a time
for row in data.itertuples():
    if row.gender == "male":
        male_count = male_count + 1
        male_total = male_total + row.GPA
    else:
        female_count = female_count + 1
        female_total = female_total + row.GPA

male_average = male_total / male_count
female_average = female_total / female_count

print("Male students:", male_count)
print("Female students:", female_count)
print("Average male grade point average:", male_average)
print("Average female grade point average:", female_average)
```
