```python
import numpy as np

FILE_PATH = r"C:\Users\admin\Desktop\student-work\Student project\Students data.csv"

gender = np.loadtxt(FILE_PATH, delimiter=",", skiprows=1, usecols=2, dtype=str)
scores = np.loadtxt(FILE_PATH, delimiter=",", skiprows=1, usecols=4)

male_scores = scores[gender == "male"]
female_scores = scores[gender == "female"]

male_count = 0
male_total = 0
for score in male_scores:
    male_count += 1
    male_total += score

female_count = 0
female_total = 0
for score in female_scores:
    female_count += 1
    female_total += score

print("Male students:", male_count)
print("Female students:", female_count)
print("Average male grade point average:", male_total / male_count)
print("Average female grade point average:", female_total / female_count)
```
