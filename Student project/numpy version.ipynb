```python
import numpy as np

FILE_PATH = r"C:\Users\admin\Desktop\student-work\Student project\Students data.csv"

male_scores = []
female_scores = []

file = open(FILE_PATH)
file.readline()                # skip the column names

for line in file:
    parts = line.strip().split(",")
    gender = parts[2]                # third column
    score = float(parts[4])          # fifth column (grade point average)

    if gender == "male":
        male_scores.append(score)
    else:
        female_scores.append(score)

file.close()

male_scores = np.array(male_scores)
female_scores = np.array(female_scores)

# Count and add up the male scores by hand
male_count = 0
male_total = 0
for score in male_scores:
    male_count = male_count + 1
    male_total = male_total + score
male_average = male_total / male_count

# Count and add up the female scores by hand
female_count = 0
female_total = 0
for score in female_scores:
    female_count = female_count + 1
    female_total = female_total + score
female_average = female_total / female_count

print("Male students:", male_count)
print("Female students:", female_count)
print("Average male grade point average:", male_average)
print("Average female grade point average:", female_average)
```
