```sql
FILE_PATH = r"C:\Users\admin\OneDrive - United States International University (USIU)\Desktop\DSA\students (1).csv"

male_ages = []
female_ages = []

file = open(FILE_PATH)
file.readline()                      # skip the column names

for line in file:
    name, age, gender = line.strip().split(",")

    if gender == "Male":
        male_ages.append(int(age))
    else:
        female_ages.append(int(age))

file.close()

print("Male students:", len(male_ages))
print("Female students:", len(female_ages))
print("Average male age:", sum(male_ages) / len(male_ages))
print("Average female age:", sum(female_ages) / len(female_ages))
```
