# A password setting exercise with a limited number of attempts

```python
password = "12345"
for i in range(10):
    a = input("enter a password: ")
    if a == password:
        print(f"you've found the password on attempt number ({i+1}) ")
        break
else:
    print("you've exceeded your attempts")
```

# Test units

```python
%%writefile calculator.py
def main():
    x = int(input("What's x? "))
    print("x squared is", square(x))


def square(n):
    return n * n


if __name__ == "__main__":
    main()
```

```python
%run calculator.py
```

```python
%%writefile test_calculator.py
import pytest
from calculator import square

def test_positive():
    assert square(2) == 4
    assert square(3) == 9


def test_negative():
    assert square(-2) == 4
    assert square(-3) == 9


def test_zero():
    assert square(0) == 0


def test_str():
    with pytest.raises(TypeError):
        square("cat")
```

```python
!pytest test_calculator.py -v
```

# testing strings

```python
%%writefile hello.py
def main():
    name = input("What's your name? ")
    print(hello(name))


def hello(to="world"):
    return f"hello, {to}"


if __name__ == "__main__":
    main()
```

```python
%%writefile test_hello.py
from hello import hello


def test_default():
    assert hello() == "hello, world"


def test_argument():
    assert hello("David") == "hello, David"
```

```python
!pytest test_hello.py -v
```

```python
%run hello.py
```

## Numpy

```python
import numpy as np
output = np.ones((5,5))
# print(output)

z = np.zeros((3,3))
z[1,1] = 9
#print(z)

output[1:-1, 1:-1] = z
print(output)
```

```python
a = np.full((5,5), 7)
print(a)
```

```python
# copying arrays
a = np.array([1,2,3])
b = a.copy()
b[0] = 100

print(b)
print(a)
```
