## squaring even numbers and finding highest scorer

```python
# squares of even numbers 1-50
even_squares = [n ** 2 for n in range(1, 51) if n % 2 == 0]
print("Even squares:", even_squares)

student_scores = {
    "Amara": [80, 90, 85],
    "Liam": [70, 75, 60],
    "Zara": [95, 92, 98],
}
student_averages = {s: mean(scores) for s, scores in student_scores.items()}
print("Student averages:", student_averages)
```
