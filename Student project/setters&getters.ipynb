## Setters, getters, and properties on classes

```python
class SensorReading:
    def __init__(self, value):
        self.value = value

    @property
    def value(self):
        return self._value

    @value.setter
    def value(self, new_value):
        if not (0 <= new_value <= 100):
            raise ValueError("Sensor reading must be between 0 and 100")
        self._value = new_value

sensor = SensorReading(45)
print("Initial sensor value:", sensor.value)

sensor.value = 99
print("Updated sensor value:", sensor.value)

try:
    sensor.value = 150  # should fail
except ValueError as e:
    print("Caught expected error:", e)
```
