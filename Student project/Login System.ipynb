```python
FILE_NAME = "users.txt"

while True:
    print("\n1. Register")
    print("2. Login")
    print("3. Quit")
    choice = input("Choose: ").strip()

    if choice == "1":
        name = input("Full name: ").strip()
        email = input("Email: ").strip().lower()
        password = input("Password: ")
        confirm = input("Confirm password: ")

        # Check if the email is already saved
        already_exists = False
        try:
            file = open(FILE_NAME)
            for line in file:
                parts = line.strip().split(",")
                if len(parts) != 3:          # skip blank or broken lines
                    continue
                if parts[1] == email:
                    already_exists = True
                    break
            file.close()
        except FileNotFoundError:
            pass

        if "@" not in email:
            print("Invalid email.")
        elif already_exists:
            print("That email is already registered. Please log in.")
        elif password != confirm:
            print("Passwords do not match.")
        else:
            file = open(FILE_NAME, "a")
            file.write(name + "," + email + "," + password + "\n")
            file.close()
            print("Account created. Welcome,", name)
            break                            # quit after registering

    elif choice == "2":
        email = input("Email: ").strip().lower()
        password = input("Password: ")
        logged_in = False

        try:
            file = open(FILE_NAME)
            for line in file:
                parts = line.strip().split(",")
                if len(parts) != 3:          # skip blank or broken lines
                    continue
                saved_name, saved_email, saved_password = parts
                if saved_email == email and saved_password == password:
                    print("Login successful. Welcome back,", saved_name)
                    logged_in = True
                    break                    # stop searching
            file.close()
        except FileNotFoundError:
            pass

        if logged_in:
            break                            # quit after logging in
        else:
            print("Wrong email or password.")

    elif choice == "3":
        print("Goodbye!")
        break

    else:
        print("Please type 1, 2 or 3.")
```
