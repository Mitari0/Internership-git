```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
data = pd.read_csv('Students data.csv')

sns.boxplot(x='gender', y='GPA', data=data)
```

```python
sns.jointplot(data=data, x="Algebra", y="GPA", kind='hex', color='blue')
plt.show()
```

```python
sns.kdeplot(data['GPA'])
```

```python

tips_data = sns.load_dataset("tips")
sns.pairplot(tips_data, hue='sex', palette='Blues')
```

## some seaborn styling 

```python
sns.set_style('ticks')
sns.set_context('paper', font_scale=1.5)

g = sns.jointplot(data=data, x="Algebra", y="GPA", kind='hex', color='blue', height=6)

sns.despine(fig=g.figure, left=True, bottom=True)
plt.show()
```

## Categorical plots

```python
## Bar Plot

sns.barplot(x='gender', y='GPA', data=data, estimator=np.std)
plt.show()
```

```python
# count plot
sns.countplot(x='gender', data=data)
```

```python
# box plot
sns.boxplot(x='class', y='GPA', data=data, hue='gender')
```

```python
# violin plot
sns.violinplot(x='class', y='GPA', data=data, hue='gender')
```

```python
# strip plot
plt.figure(figsize=(8,5))
sns.stripplot(x='class', y='GPA', data=data, hue='gender', jitter=True, dodge=True)
```

```python
# swarm plot
sns.violinplot(x='class', y='GPA', data=data)
sns.swarmplot(x='class', y='GPA', data=data, color='white')
```

```python
# palettes
plt.figure(figsize=(8,5))
sns.set_style('dark')
sns.set_context('talk')
sns.stripplot(x='class', y='GPA', data=data, hue='gender', palette='afmhot')
plt.legend(loc=2)
```

## Matrix plots

```python
plt.figure(figsize=(8, 6))
sns.set_context('paper', font_scale=1.4)

student_matrix = data.corr(numeric_only=True)

sns.heatmap(student_matrix, annot=True, cmap='Blues')
plt.show()
```

```python
flights = sns.load_dataset('flights')
flights = flights.pivot_table(index='month', columns='year', values='passengers')
sns.heatmap(flights,cmap='Blues', linewidths=1, linecolor='white')
```

```python
# cluster map
#iris= sns.load_dataset('iris')
#species=iris.pop('species')
#sns.clustermap(iris)

sns.clustermap(flights,cmap='Blues', standard_scale=1)
```

```python
# pair grid
iris= sns.load_dataset('iris')
iris_g = sns.PairGrid(iris, hue="species")
iris_g.map(plt.scatter)
iris_g.map_lower(sns.kdeplot)
```

## Regression plot


```python
sns.set_context('paper', font_scale=1.4)

sns.lmplot(x='Algebra', y='GPA', data=data, hue='gender',
           markers=['o', '^'],
           scatter_kws={'s': 100, 'linewidths': 1, 'edgecolor': 'white'},
           height=6, aspect=1.3)
plt.show()
```

```python
sns.lmplot(x='Algebra', y='GPA', data=data, col='gender', row='class')
```
